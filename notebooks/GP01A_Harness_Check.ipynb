## Cell 3 — Locate the three core harness wheels and fingerprint the actual builds


```python
CORE_PACKAGES = {'swegemma', 'adk-submission', 'adk-eval-core'}
MAX_CORE_WHEEL_BYTES = 64 * 1024 * 1024
search_root = Path(WHEELHOUSE_ROOT) if WHEELHOUSE_ROOT is not None else INPUT_ROOT
wheel_paths = sorted({p.resolve() for p in search_root.rglob('*.whl') if p.is_file()}) if search_root.is_dir() else []
CORE_WHEELS, WHEEL_RECORDS = [], []

for path in wheel_paths:
    if not path.name.lower().startswith(('swegemma-', 'adk_submission-', 'adk_eval_core-')):
        continue
    if path.stat().st_size > MAX_CORE_WHEEL_BYTES:
        HOLDS.append('Core wheel exceeds the inspection size limit: ' + path.name)
        continue
    try:
        with zipfile.ZipFile(path) as wheel:
            metadata_names = [n for n in wheel.namelist() if n.endswith('.dist-info/METADATA')]
            if len(metadata_names) != 1:
                raise ValueError('Expected one distribution METADATA file')
            member = wheel.getinfo(metadata_names[0])
            if member.file_size > 2 * 1024 * 1024:
                raise ValueError('Distribution METADATA exceeds 2 MiB')
            metadata = Parser().parsestr(wheel.read(member).decode('utf-8'))
        distribution = metadata.get('Name', '').lower().replace('_', '-')
        if distribution not in CORE_PACKAGES:
            raise ValueError('Unexpected core distribution name: ' + distribution)
        record = {
            'distribution': distribution, 'version': metadata.get('Version'), 'path': str(path),
            'bytes': path.stat().st_size, 'sha256': sha256_file(path),
            'requires_python': metadata.get('Requires-Python'),
            'requires_dist': metadata.get_all('Requires-Dist', []),
            'license_expression': metadata.get('License-Expression'),
        }
        WHEEL_RECORDS.append(record)
        CORE_WHEELS.append((path, record))
    except (OSError, ValueError, UnicodeError, zipfile.BadZipFile) as exc:
        HOLDS.append('Could not inspect ' + path.name + ': ' + str(exc))

for distribution in sorted(CORE_PACKAGES):
    builds = [r for r in WHEEL_RECORDS if r['distribution'] == distribution]
    distinct = {r['sha256'] for r in builds}
    if not builds:
        HOLDS.append('Missing ' + distribution + ' wheel. Attach the metric/Gemma 4 Developer Agent Wheelhouse dataset.')
    elif len(distinct) > 1:
        HOLDS.append('Multiple builds of ' + distribution + '; set WHEELHOUSE_ROOT to one intended release and rerun from Cell 1.')

write_json('02_core_wheels.json', WHEEL_RECORDS)
observe(3, 'Read wheel metadata and file hashes without importing the harness.',
        wheels_scanned=len(wheel_paths),
        core_builds=[{k:r[k] for k in ('distribution', 'version', 'sha256')} for r in WHEEL_RECORDS],
        holds=len(HOLDS))
```

**Observation:** Expect a captured build for each of the three packages. Package versions alone do not identify a build; SHA-256 hashes are retained. Missing or conflicting builds remain review holds.


## Cell 4 — Capture verifier, sandbox and configuration source as text


```python
MAX_TEXT_MEMBER = 2 * 1024 * 1024
MAX_TEXT_PER_WHEEL = 32 * 1024 * 1024
TEXT_SUFFIXES = {'.py', '.sh', '.yaml', '.yml', '.json', '.toml', '.ini', '.cfg', '.txt', '.md'}
SPECIAL_TEXT = {'METADATA', 'WHEEL', 'LICENSE', 'LICENSE.txt', 'Dockerfile', 'Dockerfile.sandbox', 'Dockerfile.public'}

def capture_text(relative, payload, origin):
    relative = PurePosixPath(relative)
    if relative.is_absolute() or '..' in relative.parts or '\\' in str(relative):
        raise ValueError('Unsafe source path: ' + str(relative))
    text = payload.decode('utf-8')
    destination = OUT / 'source' / Path(*relative.parts)
    destination.parent.mkdir(parents=True, exist_ok=True)
    destination.write_bytes(payload)
    row = {'file': str(destination.relative_to(OUT)), 'origin': origin,
           'bytes': len(payload), 'sha256': hashlib.sha256(payload).hexdigest()}
    SOURCE_ROWS.append(row)
    if relative.suffix == '.py':
        try:
            tree = ast.parse(text)
            for node in tree.body:
                if isinstance(node, (ast.FunctionDef, ast.AsyncFunctionDef)):
                    API_INDEX.append({'file': row['file'], 'kind': 'function', 'name': node.name,
                                      'line': node.lineno, 'signature': node.name + '(' + ast.unparse(node.args) + ')'})
                elif isinstance(node, ast.ClassDef):
                    fields = [{'name': ast.unparse(n.target), 'annotation': ast.unparse(n.annotation)}
                              for n in node.body if isinstance(n, ast.AnnAssign)]
                    API_INDEX.append({'file': row['file'], 'kind': 'class', 'name': node.name,
                                      'line': node.lineno, 'fields': fields})
                    for method in node.body:
                        if isinstance(method, (ast.FunctionDef, ast.AsyncFunctionDef)):
                            API_INDEX.append({'file': row['file'], 'kind': 'method', 'name': node.name + '.' + method.name,
                                              'line': method.lineno, 'signature': method.name + '(' + ast.unparse(method.args) + ')'})
        except SyntaxError as exc:
            HOLDS.append('Source parsing needs review: ' + row['file'] + ': ' + str(exc))

seen_builds = set()
for path, record in CORE_WHEELS:
    if record['sha256'] in seen_builds:
        continue
    seen_builds.add(record['sha256'])
    used_bytes = 0
    try:
        with zipfile.ZipFile(path) as wheel:
            for member in wheel.infolist():
                item = PurePosixPath(member.filename)
                if member.is_dir() or not (item.suffix in TEXT_SUFFIXES or item.name in SPECIAL_TEXT):
                    continue
                if member.file_size > MAX_TEXT_MEMBER or used_bytes + member.file_size > MAX_TEXT_PER_WHEEL:
                    HOLDS.append('Source capture size limit: ' + path.name + '/' + member.filename)
                    continue
                used_bytes += member.file_size
                prefix = record['distribution'] + '/' + record['sha256'][:12]
                try:
                    capture_text(prefix + '/' + member.filename, wheel.read(member), path.name)
                except (ValueError, UnicodeError) as exc:
                    HOLDS.append(str(exc))
    except (OSError, zipfile.BadZipFile) as exc:
        HOLDS.append('Wheel source read error: ' + path.name + ': ' + str(exc))

for relative in ['HARNESS_README.md', 'sandbox/setup.py', 'docker/Dockerfile.sandbox',
                 'docker/Dockerfile.public', 'docker/imp.py', 'docker/telnetlib.py']:
    path = COMPETITION_ROOT / relative
    if path.is_file() and path.stat().st_size <= MAX_TEXT_MEMBER:
        capture_text('competition_support/' + relative, path.read_bytes(), relative)
    elif relative in ('HARNESS_README.md', 'sandbox/setup.py'):
        HOLDS.append('Missing required competition support file: ' + relative)

readme_path = COMPETITION_ROOT / 'HARNESS_README.md'
readme_hash = sha256_file(readme_path) if readme_path.is_file() else None
if readme_hash != '445000219b803942c469391ec8b69791a31e9cec2f934738c767e34b5ca0aa1f':
    HOLDS.append('HARNESS_README.md differs from GP00 or is missing; review the release change.')
verifier_symbols = [r for r in API_INDEX if r['name'].split('.')[-1] == 'verify_task']
if not verifier_symbols:
    HOLDS.append('verify_task definition was not found; inspect the actual verifier entry point before GP01B.')
write_json('03_source_manifest.json', SOURCE_ROWS)
write_json('04_api_index.json', API_INDEX)
observe(4, 'Captured source and static signatures. No captured Python was imported or executed.',
        source_files=len(SOURCE_ROWS), verifier_definitions=verifier_symbols, holds=len(HOLDS))
```

**Observation:** The output should identify the actual `verify_task` definition and its signature. The returned source also records test selection, JUnit rules, sandbox setup, and protected-file handling for review.
